```python
from main import load_data

# imports
import pandas as pd
import matplotlib.pyplot as plt

bean = load_data()
bean
```

```python
print(bean.info())
bean.describe()
```

```python
# buscar valores repetidos en el dataset e imprimir los repetidos (si hay)
repeated_values = bean[bean.duplicated()]
print(repeated_values)

# eliminar los valores repetidos del dataset (si hay)
bean = bean.drop_duplicates()
bean
```

```python
# boxplots para identificar outliers en el dataset
bean = bean.drop(columns=["Class"])
# bean estandarizado para que los boxplots sean más comparables entre sí
bean_standarized = (bean - bean.mean()) / bean.std()

# junto todos los boxplots en una sola figura para tener una mejor visualización
plt.figure(figsize=(15, 10))
bean_standarized.boxplot()
plt.show()
```

```python
"""
    Como parece que "Solidity" tiene un outlier muy grande, voy a hacer un boxplot
    solo de esa columna para ver si es un outlier real o si es un error de captura de datos.
"""

plt.figure(figsize=(5, 10))
plt.boxplot(bean["Solidity"])
plt.title("Boxplot of Solidity")
plt.ylabel("Value")
plt.show()
```

Analizando los boxplots, no hay puntos que no pertenezcan al rango que maneja cada feature. A pesar de que en algunas columnas se observan puntos un poco más alejados al resto, no son extremos y aún siguen el comportamiento del resto de puntos observados; como es el caso de Solidity, en donde presenta un outlier muy grande, pero no varía demasiado del resto de datos. Por lo tanto, no se está considerando la existencia de outliers.
